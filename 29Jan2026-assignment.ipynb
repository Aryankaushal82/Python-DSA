```python
# 232. Implement Queue using Stacks
```

```python
class MyQueue(object):

    def __init__(self):
        self.stack1=[]
        self.stack2=[]

    def push(self, x):
        self.stack1.append(x)
        """
        :type x: int
        :rtype: None
        """
        
    def pop(self):
        
        while(len(self.stack1)!=0):
           self.stack2.append(self.stack1[-1])
           self.stack1.pop(-1)
        temp = self.stack2[-1]
        self.stack2.pop(-1)
        while(len(self.stack2)!=0):
            self.stack1.append(self.stack2[-1])
            self.stack2.pop(-1)
        return temp
        """
        :rtype: int
        """
        

    def peek(self):
       
        while(len(self.stack1)!=0):
           self.stack2.append(self.stack1[-1])
           self.stack1.pop(-1)
        temp = self.stack2[-1]
        while(len(self.stack2)!=0):
            self.stack1.append(self.stack2[-1])
            self.stack2.pop(-1)
        return temp
        """
        :rtype: int
        """
        

    def empty(self):
        if(len(self.stack1)==0):
            return True
        return False
        """
        :rtype: bool
        """
 
```

```python
# 225. Implement Stack using Queues
```

```python
from collections import deque

class MyStack(object):

    def __init__(self):
        self.q1 = deque()
        self.q2 = deque()

    def push(self, x):
        self.q2.append(x)
        while self.q1:
            self.q2.append(self.q1.popleft())
        self.q1, self.q2 = self.q2, self.q1

    def pop(self):
        return self.q1.popleft()

    def top(self):
        return self.q1[0]

    def empty(self):
        return not self.q1
```

```python
# 150. Evaluate Reverse Polish Notation
```

```python
class Solution:
    def evalRPN(self, tokens):
        ans = 0
        st = []

        for i in range(len(tokens)):
            if tokens[i] == "+":
                a = st[-1]
                st.pop()
                b = st[-1]
                st.pop()
                st.append(a + b)

            elif tokens[i] == "-":
                a = st[-1]
                st.pop()
                b = st[-1]
                st.pop()
                st.append(b - a)

            elif tokens[i] == "*":
                a = st[-1]
                st.pop()
                b = st[-1]
                st.pop()
                st.append(a * b)

            elif tokens[i] == "/":
                a = st[-1]
                st.pop()
                b = st[-1]
                st.pop()
                st.append(int(b / a))

            else:
                st.append(int(tokens[i]))

        if st:
            return st[-1]
        return -1
```

```python
# 71. Simplify Path
```

```python
class Solution:
    def simplifyPath(self, path: str) -> str:
        st = []
        values = path.split("/")

        for i in range(len(values)):
            val = values[i]
            if val == ".." and len(st) != 0:
                st.pop()
            elif val != "" and val != "." and val != "..":
                st.append(val)

        if len(st) == 0:
            return "/"

        ans = ""
        while len(st) != 0:
            ans = st[-1] + "/" + ans
            st.pop()
        return "/" + ans[:-1]
```

```python
# 20. Valid Parentheses
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack=[]
        for i in range (len(s)):
            if(s[i]=='(' or  s[i]=='{' or s[i]=='['):
                stack.append(s[i])
            else:
                ch = s[i]
                if(len(stack)==0):return False
                if((ch==')' and stack[len(stack)-1]=='(') or (ch==']' and stack[len(stack)-1]=='[') or (ch=='}' and stack[len(stack)-1]=='{')):
                    stack.pop()
                else:
                    return False
        
        return True
```

```python
# 2073. Time Needed to Buy Tickets
```

```python
class Solution:
    def timeRequiredToBuy(self, tickets, k):
        n = len(tickets)
        i = 0
        count = 0

        while tickets[k] != 0:
            if tickets[i % n] != 0:
                tickets[i % n] -= 1
                count += 1
            i += 1

        return count
```

```python
#  115. Distinct Subsequences  (Brute Force Without DP)
```

```python
class Solution:
    def numDistinct(self, s: str, t: str) -> int:

        def solve(i, j):
            if j < 0:
                return 1
            if i < 0:
                return 0

            if s[i] == t[j]:
                return solve(i - 1, j - 1) + solve(i - 1, j)
            else:
                return solve(i - 1, j)

        return solve(len(s) - 1, len(t) - 1)
```
